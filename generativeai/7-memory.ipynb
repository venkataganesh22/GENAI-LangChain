### Agent with Memory

```python
import os
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
```

```python
from  langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
# the above checkpoint saver stores checkpoints in memory using a defaultdict
```

```python
agent = create_agent(
    model,
    checkpointer=InMemorySaver()
) 
```

```python
config = {"configurable":{"thread_id":"test1"}}
```

```python
from langchain.messages import HumanMessage,SystemMessage 
agent.invoke({"messages":[HumanMessage(content="Hi my name is Venkat")]},config=config)
```

```python
agent.invoke({"messages":[HumanMessage(content="Hi what is my name?")]},config=config)
```

```python
### Run with Thread Id
config1 = {"configurable":{"thread_id":"test2"}}
from langchain.messages import HumanMessage,SystemMessage 
response = agent.invoke({"messages":[HumanMessage(content="Hi my name is ALice")]},config=config1)
response
```
