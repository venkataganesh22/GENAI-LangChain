**Streaming and Batch**

Streaming :
- Most of the models can stream their output content while it's being generated.By displaying output progressively,streaming significantly increases user's experience , especially for longer responses
- calling stream() return's an iterator which yeilds output chunks as they are produced .You can use loop to process those chunks in real-time

```python
for chunk in model.stream("Write 500 words about burj khalifa"):
    print(chunk.text,end="|",flush=True)
          
```

Batch :
- Batching a collection of independent requests to model can significantly improve performance and reduce cost, as the processing is done in parallel

```python
responses = model.batch(
    [
        "why is parrot green color?",
        "why does the sun rise in the east?",
        "How do planes fly?",
    ]
)
for response in responses:
    print(response)
    
```

You can also set the value of number of parallel calls or queries as shown below 

```python
responses = model.batch(
    [
        "why is parrot green color?",
        "why does the sun rise in the east?",
        "How do planes fly?",
    ],
    config={
        'max_concurrency' : 5,  #limit to 5 parallel calls
    }
)
```
