### Document loaders 

The document loaders are components in langchain used to load data from various sources into a standardized format (Document object) , which can then be used for chunking,embedding,retreival and generation.

The document no matter what format the load() function converts them into a python list and then loads them 

Let us now see different types of document loaders 

### 1. Text Loader

It loads the simple text (.txt) file 

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader('cricket.txt',encoding='utf-8')

docs = loader.load()

print(type(docs))

print(docs[0])

# print(docs[0].page_content)  for the text or data 
# print(docs[0].metadata)  for the meta data of the file

```

## 2. PyPDF Loader

It is a document loader in langchain used to load content from PDF files and convert each page into a Document object. 

If your pdf consists of 25 pages this loader returns 25 Document objects which has  page_content and metadata

- `Limitation`: Not great with scanned PDF's or complex layouts



```python
from langchain_community.document_loaders import PyPDFLoader

# install PYPDF library using pip install pypdf
%pip install pypdf

loader = PyPDFLoader("cricket_pdf.pdf")

docs = loader.load()

print(docs[0])
```

## 3. Directory Loader

It is a document loader that lets you load multiple documents from a directory of files.


```python
from langchain_community.document_loaders import DirectoryLoader,PyPDFLoader

loader = DirectoryLoader(
    path = 'directory_name',
    glob ='give the pattern eg. "*.pdf" ',
    loader_cls=PyPDFLoader
)

docs = loader.load()
```

If there are multiple files or pdf with too many pages the load time is too much and the memory might not be sufficient
 
So this load time problem can be solved using LazyLoad

### `load()` vs `lazy_load()`

* **`load()`**: Eagerly loads all documents or pages into memory (RAM) simultaneously, returning a complete list. This is ideal for smaller files or quick scripts where immediate access to the full dataset size is required.


* **`lazy_load()`**: Streams documents one by one using a Python generator, processing a single page or file at a time. This approach prevents out-of-memory errors when working with massive document collections or large directories.


```python
from langchain_community.document_loaders import DirectoryLoader,PyPDFLoader

loader = DirectoryLoader(
    path = 'directory_name',
    glob ='give the pattern eg. "*.pdf" ',
    loader_cls=PyPDFLoader
)

docs = loader.lazy_load()

for document in docs:
    print(document.page_content)
```

### 4. WebBaseLoader

It is a document loader in langchain used to load and extract text content from web pages (URL's)

It uses BeautifulSoup under the hood to parse HTML and extract visible text.

- `When to use` : for blogs,articles,or public websites where content is static.
- `Limitations` : Loads only static content not what loads after the page renders.

```python
from langchain_community.document_loaders import WebBaseLoader

url ="https://www.amazon.com/ECOLEAP-Kids-Electric-Dirt-Motor/dp/B0H33VL66J/ref=sr_1_1?_encoding=UTF8&sr=8-1"

#must install beautifulsoup library to parse html "%pip install beautifulsoup4 lxml"

loader = WebBaseLoader(url)

docs = loader.load()

print(len(docs))

print(docs[0].page_content)
```

### 5. CSV Loader 

It is a document loader used to load CSV files into langchain Document Objects one per row,by default.

```python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(file_path="cricket_formats.csv")

docs= loader.load()

print(docs[0])

print(len(docs))  #for every row in your data you get 1 document object.
```
